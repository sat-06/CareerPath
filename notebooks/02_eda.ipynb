```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\HP\Downloads\Placement_Data_Full_Class.csv")
```

```python
df.head()
```

```python
df['status'].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='status', data=df)

plt.title('Placed vs Not Placed')
plt.xlabel('Placement Status')
plt.ylabel('Number of Students')
plt.show()
```

```python
df.columns
```

```python
hist_list = ['ssc_p', 'hsc_p', 'degree_p', 'etest_p', 'mba_p', 'salary']

for col in hist_list:
    plt.figure(figsize=(6,4))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```
